```python
from PIL import Image, ImageDraw, ImageFont, ImageFilter
import PIL.ImageFont

# Create a new image with white background
image = Image.new('RGB', (480, 320), 'black')
draw = ImageDraw.Draw(image)

BackgroundColor = '#FFFFFF'
TextColor = '#000000'
HighlightColor = '#E56F39'
TintedBackgroundColor = '#F0C0A0'

if False:
    # Dark mode
    BackgroundColor = '#000000'
    TextColor = '#FFFFFF'
    HighlightColor = '#E56F39'


def drawHomeScreen(draw:ImageDraw.Draw, menuPos):
    titleFont = PIL.ImageFont.truetype('Menlo-Regular.ttf', 40)
    buttonFont = PIL.ImageFont.truetype('Menlo-Regular.ttf', 30)
    footerFont = PIL.ImageFont.truetype('Menlo-Regular.ttf', 30)

    # Background
    draw.rectangle((0, 0, 480, 320), fill=BackgroundColor)

    # Title block: Height 60
    
    # button block: Height 200, 3 x 2, Width 480
    # margin 10
    # button width = 140
    # button height = 95
    for x in range(3):
        for y in range(2):
            totalPaddingX = 15 * (x+1)
            totalPaddingY = 10 * (y)

            draw.rounded_rectangle((
                x * 140 + totalPaddingX, 
                60 + y * 95 + totalPaddingY, 
                (x + 1) * 140 + totalPaddingX, 
                60 + (y + 1) * 95 + totalPaddingY
            ), 10, 
                fill= TintedBackgroundColor, #HighlightColor if menuPos == x + y * 3 else TintedBackgroundColor,
                outline= HighlightColor if menuPos == x + y * 3 else TextColor, width=5 if menuPos == x + y * 3 else 2
            )

            
            
    # Footer block: Height 60
    draw.text((10, 260), f'{0.1234:.2f}A > {0.1234:.2f}A', fill=TextColor, font=footerFont)
            

drawHomeScreen(draw, 0)

# Display the image
display(image)
```
